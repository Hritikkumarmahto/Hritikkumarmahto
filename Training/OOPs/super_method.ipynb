# super method and MRO

- super() - Use to call parent class methods inside child class to extend parent behaviour insted of completely overriding


```python
class person:
  def __init__(self,name,city,number):
    self.name=name
    self.city=city
    self.number=number

class student(person):
  def __init__(self,name,roll_no):
    super().__init__(name)
    self.roll_no=roll_no

class common_people(student,person):
  def __init__(self,name,roll_no,city,number):
    super().__init__(name,city,number,roll_no)

# s=student("hritik",121)
# print(s.name)
# print(s.roll_no)

comm=common_people("hritik",122,"bihar",122)
print(comm.name,comm.roll_no,comm.city,comm.number)



```

```python
class meta:
  def __init__(self,user_email,password):
    self.user_email=user_email
    self.password=password

class facebook(meta):
  def __init__(self,user_email,password,user_name,profile):
    super().__init__(user_email,password)
    self.user_name=user_name
    self.profile=profile

class whatsa(meta):
  def __init__(self,user_email,password,user_name,profile,number):
    super().__init__(user_email,password)
    self.number=number

w=whatsa("Hritik@gmail.com",12323,"hritik","active",23432)

print(w.user_email)
```
