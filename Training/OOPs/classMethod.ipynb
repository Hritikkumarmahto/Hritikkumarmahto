# class Method and static method

```python
# class method 

class student:
  academy_name="Dinda academy"
  player_name="Ms dhioni"

  @classmethod
  def changeAcademyName(cls,new_name):
    cls.academy_name=new_name

  @classmethod
  def chagePlayerName(cls,new_player_name):
    cls.player_name=new_player_name


student.chagePlayerName("hritik")
student.changeAcademyName("hii viart")
print(student.academy_name,student.player_name)
```

```python
class student:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  @classmethod
  def from_string(cls,data_string):
    name,age=data_string.split(",")
    return cls(name,int(age))

s1=student.from_string("Hritik,23")  # splitting the name entered by user
print(s1.name,s1.age)
```
