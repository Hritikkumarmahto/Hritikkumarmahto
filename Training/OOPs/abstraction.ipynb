```python
from abc import ABC, abstractclassmethod

class Payment:
  @abstractclassmethod ## used abstarct which hides the implementation and shows only the functionality
  def pay(self,ammount):
    pass

class credit_Card_payment(Payment):
  def pay(self,ammount):
    print(f"paid {ammount} using credit card")

class upi_payment(Payment):
  def pay(self,ammount):
    print(f"paid {ammount} using upi")

p=credit_Card_payment()
p.pay(100)

```

Abstract class :- a calss you can not create object from directly , it is a blueprint/template for other class

Abstract Method :- A method inside that calass with no implementation


  

```python
class RBI(ABC):
  def details(self):
    print("Welcome to RBI")

  @abstractclassmethod
  def security(self):
    pass

class Sbi(RBI):
  def details(self):
   print("hii this is SBI")
   super.details()

  def security(self):
    pass

s=Sbi()
```
