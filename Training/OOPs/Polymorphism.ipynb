```python
# POlymorphish
# 1. Method Overriding

class payment:
  def pay(self,ammount):
    print(f"Genric payment method {ammount}")
class upi_payment:
  def pay(self,ammount):
    print(f"{ammount} using upi to pay")
class card_payment:
  def pay(self,ammount):
    print(f"{ammount} paying using card")


for method in [upi_payment(),card_payment(),payment()]:
  method.pay(1000)
  
```

## Overloading - When a same method name have diffrent number or types  of argument so it behave diffrently.

- Python doesn't support 100% overloading  means not fully achievable but can be achiev partially

```python
class calculator:
  def add(delf,a,b):
    return a+b
  def add(self,a,b,c):
    return a+b+c

cal=calculator()
# cal.add(1,2)    # these wont work because python doesnt support 100 % overloading 
cal.add(1,2,3)    # here the method is also overriding 
```

# we can use this using **kwargs

```python
class calculator:
  def add(delf,a,b):
    return a+b
  def add(self,a,b,c=0):
    return a+b+c

cal=calculator()
print(cal.add(1,2) )    # calling function setting c= 0  
cal.add(1,2,3)   
```

```python
# using *args we can achiev polymorphish and can give any number of argument 

class calculator:
  def add(self,*args):
    return sum(args)


cal=calculator()
print(cal.add(1,2) )    
cal.add(1,2,3)  
```

```python
# type checking

class printer:
  def show(self,value):
    if isinstance(value,str):
      print("this is a string")
    elif isinstance(value,int):
      print("this is int")
    else:
      print("method not defined")

p=printer()
p.show(12)
p.show("hritik")
```
